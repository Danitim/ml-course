## Exercise 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N
$$

$$
y = x^Tx = \sum_i{x_i^2} \implies \frac{d}{dx_k}(x^Tx) = \frac{d}{dx_k}\sum_i{x_i^2} = \frac{d}{dx_k}x_k^2 = 2x_k
$$
$$
\frac{dy}{dx} = \frac{d}{dx}(x^Tx) = 2x
$$

Согласно *Convention 3* из [Matrix Differentiation (EN)](http://www.atmos.washington.edu/~dennis/MatrixCalculus.pdf) результатом такого дифференцирования должен быть вектор-строка (т.е. транспонированный вектор) и должно получиться $2x^T$. Но так как это курс по машинному обучению, где под дифференцированием матриц чаще подразумевается их градиент, чем их матрица Якоби, то пусть результатом будет $2x$.

Classic matrix factorization example. Given matrix $X$ you need to find $A$, $S$ to approximate $X$. This can be done by simple gradient descent iteratively alternating $A$ and $S$ updates.
$$
J = || X - AS ||_F^2  , \quad A\in \mathbb{R}^{N \times R} , \quad S\in \mathbb{R}^{R \times M}
$$
$$
\frac{dJ}{dS} = ?

#### First approach

### Question 2

We can also use other distance metrics such as L1 distance.
For pixel values $p_{ij}^{(k)}$ at location $(i,j)$ of some image $I_k$,

the mean $\mu$ across all pixels over all images is $$\mu=\frac{1}{nhw}\sum_{k=1}^n\sum_{i=1}^{h}\sum_{j=1}^{w}p_{ij}^{(k)}$$
And the pixel-wise mean $\mu_{ij}$ across all images is
$$\mu_{ij}=\frac{1}{n}\sum_{k=1}^np_{ij}^{(k)}.$$
The general standard deviation $\sigma$ and pixel-wise standard deviation $\sigma_{ij}$ is defined similarly.

Which of the following preprocessing steps will not change the performance of a Nearest Neighbor classifier that uses L1 distance? Select all that apply.
1. Subtracting the mean $\mu$ ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu$.)
2. Subtracting the per pixel mean $\mu_{ij}$  ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu_{ij}$.)
3. Subtracting the mean $\mu$ and dividing by the standard deviation $\sigma$.
4. Subtracting the pixel-wise mean $\mu_{ij}$ and dividing by the pixel-wise standard deviation $\sigma_{ij}$.
5. Rotating the coordinate axes of the data.

*Your Answer:*

Преобразования **1**, **2** и **3** не поменяют работу kNN.

*Your Explanation:*

Рассмотрим как поменяются расстояния при данных преобразованиях:

1) $|\tilde{x}_{ij} - \tilde{y}_{ij}| = |(x_{ij} - \mu) - (y_{ij} - \mu)| = |x_{ij} - y_{ij}|$
2) $|\tilde{x}_{ij} - \tilde{y}_{ij}| = |(x_{ij} - \mu_{ij}) - (y_{ij} - \mu_{ij})| = |x_{ij} - y_{ij}|$
3) $|\tilde{x}_{ij} - \tilde{y}_{ij}| = \frac{1}{\sigma}|x_{ij} - y_{ij}|$. Равномерное сжатие всех расстояний не поменяет ранжирование алгоритма kNN.
4) $|\tilde{x}_{ij} - \tilde{y}_{ij}| = |\frac{x_{ij} - y_{ij}}{\sigma_{ij}}|$. Теперь каждый пиксель стал взвешиваться по-разному, а значит расстояния поменяются.
5) Так как L1 норма не инварианта к линейным преобразованиям (включая повороты осей), то расстояния до и после поворота совпадать не будут

## Question 3

Which of the following statements about $k$-Nearest Neighbor ($k$-NN) are true in a classification setting, and for all $k$? Select all that apply.
1. The decision boundary (hyperplane between classes in feature space) of the k-NN classifier is linear.
2. The training error of a 1-NN will always be lower than that of 5-NN.
3. The test error of a 1-NN will always be lower than that of a 5-NN.
4. The time needed to classify a test example with the k-NN classifier grows with the size of the training set.
5. None of the above.

*Your Answer:*

Верными являются утверждения **2** и **4**.

*Your Explanation:*

1) Это не так, потому что границы между классами $-$ это сложные области в пространстве признаков и даже при $k=1$ это нелинейные области.

4) Чтобы классифицировать точку из тестовой выборки, необходимо рассчитать все расстояния до точек из тренировочной выборки, что занимает по сложности $O(N_{\text{train}} \cdot d)$. Значит, увеличение размера обучающей выборки увеличит время классификации тестовой точки.
